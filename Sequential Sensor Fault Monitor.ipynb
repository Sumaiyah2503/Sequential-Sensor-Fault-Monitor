```python
# Create synthetic sensor data

np.random.seed(42)

data = {
    "Time": range(1, 21),
    "Temperature": [30, 32, 31, 33, 35, 34, 36, 38, 40, 42,
                    45, 48, 52, 55, 58, 60, 45, 38, 34, 32],
    "Pressure": [100, 102, 99, 101, 103, 100, 98, 97, 96, 95,
                 94, 92, 88, 85, 82, 80, 95, 98, 100, 101],
    "Vibration": [2.0, 2.1, 1.8, 2.3, 2.5, 2.2, 2.8, 3.0, 3.2, 3.5,
                  3.8, 4.0, 5.5, 6.0, 6.5, 7.0, 4.2, 3.0, 2.5, 2.1]
}

df = pd.DataFrame(data)

print("Sensor Data:")
display(df)
```

```python
# Define sensor thresholds

TEMP_LIMIT = 50
PRESSURE_MIN = 90
PRESSURE_MAX = 110
VIBRATION_LIMIT = 5

# Check each sensor condition

def check_sensor(row):
    temperature_fault = row["Temperature"] > TEMP_LIMIT
    pressure_fault = not (PRESSURE_MIN <= row["Pressure"] <= PRESSURE_MAX)
    vibration_fault = row["Vibration"] > VIBRATION_LIMIT

    if temperature_fault or pressure_fault or vibration_fault:
        return "ABNORMAL"
    else:
        return "NORMAL"

df["Sensor_Condition"] = df.apply(check_sensor, axis=1)

display(df)
```

```python
# Sequential State Machine

state = "NORMAL"
states = []

for condition in df["Sensor_Condition"]:

    if state == "NORMAL":
        if condition == "ABNORMAL":
            state = "WARNING"

    elif state == "WARNING":
        if condition == "ABNORMAL":
            state = "FAULT"
        else:
            state = "NORMAL"

    elif state == "FAULT":
        if condition == "NORMAL":
            state = "RECOVERY"

    elif state == "RECOVERY":
        if condition == "NORMAL":
            state = "NORMAL"
        else:
            state = "FAULT"

    states.append(state)

df["State"] = states

display(df)
```

```python
import numpy as np
import pandas as pd

# Define sensor thresholds (copied from CRaKttu9d3fW)
TEMP_LIMIT = 50
PRESSURE_MIN = 90
PRESSURE_MAX = 110
VIBRATION_LIMIT = 5

# Test Cases (copied from EJ5t0WVjecju)
test_cases = pd.DataFrame({
    "Test_Case": [
        "N1", "N2", "N3", "N4", "N5",
        "N6", "N7", "N8", "N9", "N10",
        "F1", "F2", "F3", "F4", "F5"
    ],

    "Temperature": [
        30, 35, 40, 45, 28,
        32, 38, 42, 48, 25,
        60, 30, 55, 40, 52
    ],

    "Pressure": [
        100, 105, 95, 102, 98,
        100, 97, 103, 108, 101,
        100, 80, 100, 120, 85
    ],

    "Vibration": [
        2.0, 2.5, 3.0, 4.0, 1.5,
        2.2, 3.5, 4.5, 4.8, 2.0,
        2.0, 2.0, 6.0, 2.0, 7.0
    ],

    "Expected_Type": [
        "NORMAL", "NORMAL", "NORMAL", "NORMAL", "NORMAL",
        "NORMAL", "NORMAL", "NORMAL", "NORMAL", "NORMAL",
        "FAULT", "FAULT", "FAULT", "FAULT", "FAULT"
    ]
})

# Verify test cases

def classify_sensor(row):
    temperature_fault = row["Temperature"] > TEMP_LIMIT
    pressure_fault = not (PRESSURE_MIN <= row["Pressure"] <= PRESSURE_MAX)
    vibration_fault = row["Vibration"] > VIBRATION_LIMIT

    if temperature_fault or pressure_fault or vibration_fault:
        return "FAULT"
    else:
        return "NORMAL"


test_cases["Actual_Result"] = test_cases.apply(classify_sensor, axis=1)

test_cases["Result"] = np.where(
    test_cases["Expected_Type"] == test_cases["Actual_Result"],
    "PASS",
    "FAIL"
)

display(test_cases)
```

```python
# STEP 8: Result Summary

total_cases = len(test_cases)
passed_cases = (test_cases["Result"] == "PASS").sum()
failed_cases = (test_cases["Result"] == "FAIL").sum()

accuracy = (passed_cases / total_cases) * 100

print("===== TEST RESULT SUMMARY =====")
print("Total Test Cases :", total_cases)
print("Passed Cases     :", passed_cases)
print("Failed Cases     :", failed_cases)
print("Accuracy         :", accuracy, "%")
```

```python
# STEP 10: State Transition Table

transition_table = pd.DataFrame({
    "Current State": [
        "NORMAL",
        "NORMAL",
        "WARNING",
        "WARNING",
        "FAULT",
        "RECOVERY"
    ],

    "Input Condition": [
        "NORMAL",
        "ABNORMAL",
        "NORMAL",
        "ABNORMAL",
        "NORMAL",
        "NORMAL"
    ],

    "Next State": [
        "NORMAL",
        "WARNING",
        "NORMAL",
        "FAULT",
        "RECOVERY",
        "NORMAL"
    ]
})

display(transition_table)
```

```python
# STEP 11: State Flow Visualization

state_order = ["NORMAL", "WARNING", "FAULT", "RECOVERY", "NORMAL"]

plt.figure(figsize=(10, 4))

plt.plot(
    range(len(state_order)),
    range(len(state_order)),
    marker="o"
)

plt.xticks(range(len(state_order)), state_order)
plt.yticks([])
plt.xlabel("Sequential State Flow")
plt.title("Sequential Sensor Fault Monitor - State Flow")
plt.grid(True)

plt.show()
```
